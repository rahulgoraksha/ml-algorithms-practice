# Matplotlib library: 

- It is comprehensive plotting library in python for creating statis, animated, and interactive visualisations.

- It provides object-oriented API for embedding plots into applications using various GUI suppports a wise range of plot types

- It oofers extensive control over plot elements, allowing users to customize colors, labels, titles, gridlines, legends, and plot style to create publication quality figures.

- It supports creating dynamic visualization and interactive plots, enabling features like zooming, panning, and animations.

```python
import numpy as numpy
import matplotlib.pyplot as plt  #importing matplotlib module
```

```python
x = [0,1,2,3,4]
y = [0,1,4,9,16]

plt.plot(x,y)  #plotting x and y values
plt.show()  #displaying the plot    
```

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8]
y = [0, 4, 16, 36, 64]

fig, ax = plt.subplots()  # creating a figure and axis object
ax.plot(x, y, marker="o", label='Sales v/s purchase')
ax.set_title('Bombay coated and special steels limited')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.legend()  # to show the label

plt.show()  # displaying the plot
   
```

```python
x = [0,1,2,3,4]
y = [0, 1,4,9,16]

#plotting x and y values
plt.plot(x,y)

# Adding title and labels
plt.title('Bombay coated and special steels limited')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Displaying the plot
plt.show()  #displaying the plot

```

# Adding marker

```python
plt.plot(x, y, marker = "o", color = 'red', linestyle = '--')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

# Multiple lines on same plot

```python
y2 = [0, 1, 8, 27, 64]
plt.plot(x, y, marker = 'o', color = 'red')
plt.plot(x , y2, marker = 's', color = 'blue')
plt.title('Multiple lines plot')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

# Bar Plot

```python
categories = ['A', 'B', 'C', 'D']
values = [10,15, 7,12]

plt.bar( categories, values, color = 'skyblue')
plt.title('Bombay Steels Category-wise values')

plt.show()
```

```python
plt.scatter(x,y , color = 'green', marker = '^')
plt.title('Scatter plot example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')    
plt.show()
```

```python
data = [1,2,2,3,3,3,4,4,5,5,5,5]

plt.hist(data, bins=5, color='orange', edgecolor='black')
plt.title('Bombay Steel Histogram')
plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
a = np.linspace(0,10,11)
b = a ** 4
```

```python
x = np.arange(0,11)
y = 2 * x
```

```python
fig,axes = plt.subplots()

# Plotting x and y values
axes.plot(x,y) 
```

```python
# Adding number of rows and columns for subplots
# Here, we are creating 1 row and 2 columns of subplots
fig, axes = plt.subplots(nrows = 1, ncols = 2 )
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 2)

# Plotting x and y values on first subplot
axes[0].plot(x,y) # Plotting a and b values on second subplot
axes[1].plot(a,b) # Plotting a and b values on second subplot
```

```python
# Adding number of rows and columns for subplots
# Here, we are creating 2 rows and 2 columns of subplots
fig, axes = plt.subplots(nrows = 2, ncols = 2)


# Plotting x and y values on first subplot
axes[0][0].plot(x,y)
# Plotting a and b  values on fourth subplot
axes[1][1].plot(a,b)

```

```python
fig, axes = plt.subplots(nrows = 3, ncols = 1)

# Looping through each axis to plot
for ax in axes:
    ax.plot(x,y)
```

```python
fig, axes = plt.subplots(nrows = 3, ncols = 1)
# Plotting x and y values on first subplot
axes[0].plot(x,y)
axes[1].plot(x,y)
axes[2].plot(x,y)

plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows = 3, ncols = 1)
# Plotting x and y values on first subplot
axes[0].plot(x,y)
axes[1].plot(x,y)
axes[2].plot(x,y)

fig.subplots_adjust(wspace=0.9, hspace=0.9)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,11,10)
```

```python
# Creating a figure and axis object
fig = plt.figure()
axes = fig.add_axes([0,0,1,1])

# Plotting x and x values on the axes
axes.plot(x,x, label='x vs x')
 
# Plotting x and x squared values on the same axes
axes.plot (x,x**2, label='x vs x squared')

axes.legend()  # Adding legend to the axes
```
