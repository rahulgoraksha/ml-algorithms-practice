# Matplotlib library: 

- It is comprehensive plotting library in python for creating statis, animated, and interactive visualisations.

- It provides object-oriented API for embedding plots into applications using various GUI suppports a wise range of plot types

- It oofers extensive control over plot elements, allowing users to customize colors, labels, titles, gridlines, legends, and plot style to create publication quality figures.

- It supports creating dynamic visualization and interactive plots, enabling features like zooming, panning, and animations.

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8]
y = [0, 4, 16, 36, 64]

fig, ax = plt.subplots()  # creating a figure and axis object
ax.plot(x, y, marker="o", label='Sales v/s purchase')
ax.set_title('Bombay coated and special steels limited')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.legend()  # to show the label

plt.show()  # displaying the plot
   
```

```python
x = [0,1,2,3,4]
y = [0, 1,4,9,16]

#plotting x and y values
plt.plot(x,y)

# Adding title and labels
plt.title('Bombay coated and special steels limited')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Displaying the plot
plt.show()  #displaying the plot

```

# Adding marker

```python
plt.plot(x, y, marker = "o", color = 'red', linestyle = '--')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

# Multiple lines on same plot

```python
y2 = [0, 1, 8, 27, 64]
plt.plot(x, y, marker = 'o', color = 'red')
plt.plot(x , y2, marker = 's', color = 'blue')
plt.title('Multiple lines plot')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

# Bar Plot

```python
categories = ['A', 'B', 'C', 'D']
values = [10,15, 7,12]

plt.bar( categories, values, color = 'skyblue')
plt.title('Bombay Steels Category-wise values')

plt.show()
```

```python
plt.scatter(x,y , color = 'green', marker = '^')
plt.title('Scatter plot example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')    
plt.show()
```

```python
data = [1,2,2,3,3,3,4,4,5,5,5,5]

plt.hist(data, bins=5, color='orange', edgecolor='black')
plt.title('Bombay Steel Histogram')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

fig = plt.figure()
gs = gridspec.GridSpec(2, 2)

ax1 = fig.add_subplot(gs[:, 0])   # Big plot (left)
ax2 = fig.add_subplot(gs[0, 1])   # Top-right
ax3 = fig.add_subplot(gs[1, 1])   # Bottom-right

ax1.plot([1,2,3])
ax2.plot([3,2,1])
ax3.plot([1,3,2])

plt.show()
```

```python
fig, axes = plt.subplots(2, 2)

axes[0,0].plot([1,2,3], [1,2,3])
axes[0,0].set_title("Top Left")

axes[0,1].plot([1,2,3], [1,4,9])
axes[0,1].set_title("Top Right")

axes[1,0].plot([1,1,1], [1,2,3])
axes[1,0].set_title("Bottom Left")

axes[1,1].plot([1,2,3], [3,2,1])
axes[1,1].set_title("Bottom Right")

plt.tight_layout()   # Adjust spacing
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2)  # 1 row, 2 columns

axes[0].plot([1,2,3], [1,4,9])
axes[0].set_title("Plot 1")

axes[1].plot([1,2,3], [1,2,3])
axes[1].set_title("Plot 2")

plt.show()
```
