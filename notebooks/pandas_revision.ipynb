Indexing in Pandas

```python
data = {'Name': ['Alice', 'Bob', 'Charlie'],'Age': [25, 30, 35]}
df = pd.DataFrame(data)

# Select a single column
df_index = df['Age']
print(df_index)

# Select multiple columns
df_multiple_index = df[['Name', 'Age']]
print("\nmultiple index: ",df_multiple_index) 

#select rows by position
row_0 = df.iloc[0]
print("\nrow 0: ",row_0)

# select by iloc for multiple rows
rows_0_1 = df.loc[0:2] 
print("\nrows 0 to 2: ",rows_0_1)
```

Handling Missing Data in Pandas

Identifying Missing Data With Pandas using:

isnull(): Returns True where values are missing (NaN), and False where values are present.
notnull(): Returns True where ues are present, and False where they are missing.

```python
import numpy as np
df = pd.DataFrame({'Col1': [1, 2, np.nan],'Col2': [3, np.nan, np.nan]})
df
```

```python
print(df.isnull()) #Returns True where values are missing (NaN), and False where values are present.
```

```python
print(df.notnull) #Returns True where ues are present, and False where they are missing.
```

Filling Missing Data

Aggregation and Grouping With Pandas

```python
import pandas as pd
data = {'Category': ['A', 'B', 'A', 'B'], 'Value': [10, 20, 30, 40]}
df = pd.DataFrame(data)
df
```

```python
# Group data by 'Category' and calculate the sum
group_data = df.groupby('Category').sum()
group_data
```

```python
# Compute the mean for each group
mean_data = df.groupby('Category').mean()
mean_data
```

```python
# Aggregate using multiple functions
aggregate_data_mutiple  = df.groupby('Category').agg(['sum', 'mean'])
print(aggregate_data_mutiple )   #.sum(), .mean(), and .size(): Perform aggregation on the grouped data.
```
