```python
#what are the top 5 most common country code in the dataset
df['country'].value_counts().head(5)
```

```python
#what is the name of the person who paid highest ADR (Average Daily rate), how much was thier ADR
result = df['adr'].max()
result

print("adr     "+"    name")
print(df[df['adr'] == result]['name'])
```

```python
#the average daily adr rate of a person staying in hotel, what a mean across the hotel satys in dataset
mean = df['adr'].mean()
mean.round(2)
```

```python
#what is the avg of the night across the stay for the entire dataset, feel free to round this 2 decimal
df['total_across_night'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['total_across_night'].mean().round(2)
```

```python
# What is the average total cost for a stay in the dataset? Not *average daily cost*, but *total* stay cost. (You will need to calculate total cost your self by using ADR and week day and weeknight stays). Feel free to round this to 2 decimal points.**
df['total_cost'] =df['adr'] * (df['stays_in_week_nights']+df['stays_in_weekend_nights'])
df['total_cost'].mean().round(2)
```

```python
#What are the names and emails of people who made exactly 5 "Special Requests"?**
print(df[df['total_of_special_requests']==5]['name'])
```

```python
#What percentage of hotel stays were classified as "repeat guests"? (Do not base this off the name of the person, but instead of the is_repeated_guest column)**
total_stays = len(df)
repeated_stay = df['is_repeated_guest'].sum()

percentage = (repeated_stay/total_stays)*100
percentage
```

```python
"""What are the top 5 most common last name in the dataset? Bonus: \
Can you figure this out in one line of pandas code?
 (For simplicity treat the a title such as MD as a last name,
  for example Caroline Conley MD can be said to have the last name MD)**"""

df['name'].str.split().str[-1].value_counts().head(5)
```

```python
#What are the names of the people who had booked the most number children and babies for their stay? (Don't worry if they canceled, only consider number of people reported at the time of their reservation)
df['total_people'] = df['children'] + df['babies']
df.sort_values('total_people',ascending=False)[['name','adults','total_people','babies','children']][:3]
```

```python
# What are the top 3 most common area code in the phone numbers? (Area code is first 3 digits)
df['phone-number'].str[:3].value_counts().head(3)
```

```python
# How many arrivals took place between the 1st and the 15th of the month (inclusive of 1 and 15) ? Bonus: Can you do this in one line of pandas code?
df['arrival_date_day_of_month'].apply(lambda day: day in range(1,16)).sum()
```
